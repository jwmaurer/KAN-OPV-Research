Purpose: To graphically analyze the top n PCE molecules in the HOPV-15 dataset

```python
from import_dataset import open_hopv15_dataset, mol_atom_dist
from rdkit import Chem
import matplotlib.pyplot as plt

data = open_hopv15_dataset("./HOPV_15_revised_2.data")
n = 10
```

```python
sorted_data = sorted(data, key=lambda x: x.experimental_data.PCE, reverse=True)
```

```python
top_n_chems = [Chem.AddHs(Chem.MolFromSmiles(molecule.smiles_molecule)) for molecule in data[:n]]
```

```python
for i, chem in enumerate(top_n_chems):
    print(mol_atom_dist(chem), sorted_data[i].experimental_data.PCE)
```

```python
from rdkit.Chem import Draw
for chem in top_n_chems:
    img = Draw.MolToImage(chem)
    print(img.height, img.width)
    # img.show()
```
