```python
import numpy as np
import torch
from molecule_autoencoder import MoleculeAutoEncoderSigmoid, create_autoencoder, create_layers
import torch.nn as nn
```

```python
models = list(range(0,99))
tanh_accuracies = []
sigmoid_accuracies = []
for model in models:
    with open("D:\\MoleculeAutoEncoder\\tanh_models\\model_"+str(model)+"\\model_"+str(model)+"_report.txt", 'r') as tanh_file:
        for line in tanh_file.readlines():
            line = line.lower().strip().split()
            if "test" in line:
                tanh_accuracies.append(float(line[-1]))
    with open("D:\\MoleculeAutoEncoder\\sigmoid_models\\model_"+str(model)+"\\model_"+str(model)+"_report.txt", 'r') as sigm_file:
        for line in sigm_file.readlines():
            line = line.lower().strip().split()
            if "test" in line:
                sigmoid_accuracies.append(float(line[-1]))
```

```python
minimum_sigmoid = None
for model in models:
    with open("D:\\MoleculeAutoEncoder\\tanh_models\\model_"+str(model)+"\\model_"+str(model)+"_report.txt", 'r') as sigm_file:
            for line in sigm_file.readlines():
                line = line.lower().strip().split()
                if "test" in line and float(line[-1]) == np.min(tanh_accuracies):
                    minimum_sigmoid = torch.load("D:\\MoleculeAutoEncoder\\sigmoid_models\\model_"+str(model)+"\\model_"+str(model)+".pt", weights_only=False)
```

```python
minimum_sigmoid
```

```python
min_encoder_str = "4|373->gelu->373|582->gelu->582|326->gelu->326|488->gelu->488|1"
min_decoder_str = "1|434->gelu->434|371->silu->371|377->gelu->377|484->relu->484|4"
```

```python
str_to_activ = {"gelu": nn.GELU(), "relu": nn.ReLU(), "silu": nn.SiLU(), "mish": nn.Mish()}
encoder_layers = create_layers(min_encoder_str, str_to_activ)
decoder_layers = create_layers(min_decoder_str, str_to_activ)
loaded_model = MoleculeAutoEncoderSigmoid(encoder_layers, decoder_layers)
loaded_model.load_state_dict(minimum_sigmoid())
```
