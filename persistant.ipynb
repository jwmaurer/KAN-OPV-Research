```python
index_mask = np.argwhere(data_mask == False).flatten()
```

```python
import csv

line_num = 0
with open("./OPV_Datasets/molecule_images.csv", 'rt') as file:
    with open("./OPV_Datasets/smaller_molecular_images.csv", 'wt') as file2:
        writer = csv.writer(file2, delimiter=',')
        for line in file:
            print(f"Line: {line_num}")
            if len(line) == 1:
                line_num += 1
                continue
            else:
                values = np.asarray(line.strip().split(','), dtype=np.int16)
                writer.writerow(values[index_mask])
            line_num += 1
```
